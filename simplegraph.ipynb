Build a simple workflow using Langgraph

```python
from typing_extensions import TypedDict
class State(TypedDict):
    graph_info:str
```

```python
def start_play(state:State):
    print("Start_Play node has been called")
    return{"graph_info":state['graph_info'] + "I am planning to play"}

def badminton(state:State):
    print("Badminton has been called")
    return{"graph_info":state['graph_info'] + "I am planning to play badminton" }

def cricket(state:State):
    print("Cricket has been called")
    return{"graph_info":state['graph_info'] + "I am planning to play cricket" }
```

```python
import random
from typing import Literal

def random_play(state:State) -> Literal['badminton','çricket']:
    graph_info = state[graph_info]
    if random.random >0.5:
        return "cricket"
    else:
        return "badminton"
```
